2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
由於 0.1 的二進位表示是無限循環小數，因此無法完全儲存。

Python 使用 IEEE 754 雙精度浮點數（Double Precision）來儲存浮點數。因為儲存空間有限，所以只能保留部分位數，其餘位數必須捨去。

當需要捨入時，IEEE 754 採用「四捨六入五成雙（Round Half To Even）」的方法，以減少長期計算時累積的誤差。

因此 Python 實際儲存的是最接近 0.1 的近似值，而不是完全精確的 0.1。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
程式：

Python
print(0.1 + 0.2)

輸出：

Plain Text
0.30000000000000004
``

原因是 0.1 與 0.2 都無法被二進位精確表示，因此實際儲存的是近似值，兩個近似值相加後便產生微小誤差。

可以使用：

Python
print(round(0.1 + 0.2, 1))

輸出：

Plain Text
0.3
 

或使用較精確的 Decimal：

Python
from decimal import Decimal
 
print(Decimal('0.1') + Decimal('0.2'))

輸出：

Plain Text
0.3
5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
如果只是使用 round()，雖然可以讓結果看起來正確，但實際上的誤差仍然存在。因此在大量計算或高精度需求的情況下，並不是最好的方法。

對於科學計算而言，通常可以接受極小的浮點數誤差，因為計算速度較重要。

但在金融計算中，金額必須非常精確，因此不建議只使用 round() 處理，而應該使用 Decimal 來避免浮點數誤差。

因此，round() 適合修正顯示結果；若需要精確計算，尤其是金融領域，使用 Decimal 會是較好的方法。
