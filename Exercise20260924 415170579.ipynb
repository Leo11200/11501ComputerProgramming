### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

VRAM 減半

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
Python 可以正常運作，因為 int 採用 Arbitrary Precision，位數不夠時會動態配置更多記憶體來儲存數字。
但 AI 模型通常有數十億到數千億個權重，若每個權重都使用任意精度或 FP64，記憶體與運算成本會高得難以接受。因此模型通常採用 FP32、FP16 或 BF16，在「精度夠用」的前提下，大幅降低記憶體用量、提高 GPU 吞吐量與訓練速度。
