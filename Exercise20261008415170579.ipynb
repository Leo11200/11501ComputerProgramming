使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。
若 n=5，最大值 5 的二進位是 101，所以左邊兩個乘數寬度都是 3 位；乘積最大為 5×5=25，二進位是 11001，所以結果寬度是 5 位。

Python
n = int(input())
 
bw = len(f"{n:b}")
pw = len(f"{n*n:b}")
 
for i in range(1, n + 1):
for j in range(1, n + 1):
print(
f"{i:0{bw}b}×{j:0{bw}b}={i*j:0{pw}b}",
end=" "
)
print()

1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。 
如果限制 只能使用 while，可以改成：

Python
n = int(input())
 
bw = len(f"{n:b}")
pw = len(f"{n*n:b}")
 
i = 1
while i <= n:
j = 1
while j <= n:
print(
f"{i:0{bw}b}×{j:0{bw}b}={i*j:0{pw}b}",
end=" "
)
j += 1
print()
i += 1
